```python
edu_order = {'High School':0, 'Bachelors':1, 'Masters':2, 'PhD':3}
df['EducationLevel'] = df['EducationLevel'].map(edu_order)
```

```python
df = pd.get_dummies(df, columns=['Department'], dtype=int)
```

```python
from sklearn.preprocessing import LabelEncoder
enco=LabelEncoder()
df['OverTime']= enco.fit_transform(df['OverTime'])
df['Attrition']=enco.fit_transform(df['Attrition'])
df.head()
```

```python
X=df.drop('Attrition',axis=1)
y=df['Attrition']
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y, test_size=0.2, random_state=2,stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_sclr=scaler.fit_transform(X_train)
X_test_sclr=scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=35)
knn.fit(X_train_sclr,y_train)
```

```python
y_pred=knn.predict(X_test_sclr)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logr=LogisticRegression()
logr.fit(X_train_sclr,y_train)
```

```python
yy=logr.predict(X_test_sclr)
print(accuracy_score(y_test,yy))
```
