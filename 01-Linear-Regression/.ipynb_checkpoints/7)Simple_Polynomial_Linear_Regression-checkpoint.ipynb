## Using simple linear regression
*Accuracy is not good*

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("cgpa_salary.csv")
df.head()
```

```python
X=df.iloc[:,0:1]
X
```

```python
y=df.iloc[:,-1]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
X_train
```

```python
X_test.size
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
```

```python
lr.predict(X_test.iloc[[1]])
```

```python
y_test
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
y_pred=lr.predict(X_test)
score=r2_score(y_test,y_pred)
score
```

```python
mse= mean_squared_error(y_test,y_pred)
mse
```

```python
mae=mean_absolute_error(y_test,y_pred)
mae
```

## Using Polynomial linear regression on same dataset

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=2)
```

```python
X_train_trans=poly.fit_transform(X_train)
X_test_trans=poly.transform(X_test)
```

```python
X_train_trans[0]
```

```python
plr=LinearRegression()
plr.fit(X_train_trans,y_train)
```

```python
y_pred=plr.predict(X_test_trans)
```

```python
y_test
```

```python
scr=r2_score(y_test,y_pred)
scr
```

```python
mse= mean_squared_error(y_test,y_pred)
mse
```

```python
mae=mean_absolute_error(y_test,y_pred)
mae
```
