```python
def perceptron(X,y):
    X=np.insert(X,0,1, axis=1)
    weight=np.ones(X.shape[1])
    learning_rate=0.1

    for i in range(1000):
        j=np.random.randint(0,100)
        y_hat=Sigmoid(np.dot(X[j], weight))
        weight=weight + learning_rate * (y[j]-y_hat)*X[j]
    return weight[0], weight[1:]
```

```python
def Sigmoid(z):
    return 1/(1+np.exp(-z))
```
